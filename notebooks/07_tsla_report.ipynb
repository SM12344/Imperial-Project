## 1. Inputs and Outputs


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent

def configure_script_path(project_root: Path) -> Path:
    scripts_dir = project_root / 'scripts'
    if str(scripts_dir) not in sys.path:
        sys.path.insert(0, str(scripts_dir))
    return scripts_dir

SCRIPTS_DIR = configure_script_path(PROJECT_ROOT)
print(f'Initial project root: {PROJECT_ROOT}')
print(f'Scripts directory: {SCRIPTS_DIR}')
```

```python
try:
    from google.colab import drive  # type: ignore
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

if IN_COLAB:
    drive.mount('/content/drive')
    COLAB_PROJECT_ROOT = Path('/content/drive/MyDrive/Project')
    PROJECT_ROOT = COLAB_PROJECT_ROOT.resolve()
    SCRIPTS_DIR = configure_script_path(PROJECT_ROOT)
    print(f'Using project root: {PROJECT_ROOT}')
    print(f'Scripts directory: {SCRIPTS_DIR}')
else:
    print('Not running in Colab. Skipping Drive mount.')
```
