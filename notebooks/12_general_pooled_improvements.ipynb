## What This Stage Tests


```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent

TABLES_DIR = PROJECT_ROOT / 'outputs' / 'tables'

holdout_results = pd.read_csv(TABLES_DIR / 'general_pooled_holdout_results.csv')
walkforward_results = pd.read_csv(TABLES_DIR / 'general_pooled_walkforward_results.csv')
walkforward_summary = pd.read_csv(TABLES_DIR / 'general_pooled_walkforward_summary.csv')
metadata = pd.read_csv(TABLES_DIR / 'general_pooled_metadata.csv')

metadata
```

## Best General Candidates


```python
best_holdout = holdout_results.sort_values(
    ['slice_name', 'roc_auc', 'balanced_accuracy', 'f1', 'accuracy'],
    ascending=[True, False, False, False, False],
).groupby('slice_name').head(3)

best_walkforward = walkforward_summary.sort_values(
    ['slice_name', 'mean_roc_auc', 'mean_balanced_accuracy', 'mean_f1', 'mean_accuracy'],
    ascending=[True, False, False, False, False],
).groupby('slice_name').head(3)

best_holdout, best_walkforward
```
