## Inspect Label Coverage


```python
coverage = (
    article_dataset.groupby("ticker")
    .agg(
        rows=("article_id", "size"),
        unique_articles=("article_id", "nunique"),
        unique_aligned_dates=("aligned_trading_date", "nunique"),
        target_up_rate=("target_next_day_up", "mean"),
        mean_finbert_sentiment=("finbert_sentiment_score", "mean"),
    )
    .reset_index()
    .sort_values("rows", ascending=False)
)

coverage
```
