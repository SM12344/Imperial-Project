# 20 Portable Regression Experiments


## Split


```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent

TABLES = PROJECT_ROOT / 'outputs' / 'tables'
SUFFIX = '2020_2024_polygon_proxy_market_cleaned_news'

def load_table(name):
    path = TABLES / f'{name}_{SUFFIX}.csv'
    return pd.read_csv(path)

early = load_table('portable_regression_early_results')
late = load_table('portable_regression_late_fusion_results')
global_blend = load_table('portable_regression_global_ticker_blend_results')
bootstrap = load_table('portable_regression_bootstrap_results')
selective = load_table('portable_regression_selective_results')
summary = load_table('portable_regression_best_summary')
overview = load_table('portable_regression_overview')

overview
```

## Best Large All-Row Results


```python
all_results = pd.concat([early, late, global_blend, bootstrap], ignore_index=True, sort=False)
large = all_results[all_results['test_rows'] >= 500].copy()
cols = [
    'experiment_family', 'scope', 'horizon', 'return_kind', 'target', 'feature_set',
    'model_label', 'blend_method', 'selected_model', 'test_rows',
    'directional_accuracy', 'directional_majority_baseline',
    'directional_accuracy_minus_baseline', 'directional_balanced_accuracy',
    'rmse', 'baseline_rmse_train_mean', 'rmse_improvement_vs_train_mean', 'r2', 'correlation'
]
cols = [c for c in cols if c in large.columns]
large.sort_values(
    ['directional_accuracy_minus_baseline', 'directional_balanced_accuracy', 'directional_accuracy'],
    ascending=[False, False, False]
)[cols].head(20)
```

## Best Large Selective Results


```python
large_selective = selective[selective['test_covered_rows'] >= 500].copy()
sel_cols = [
    'experiment_family', 'scope', 'horizon', 'return_kind', 'target', 'blend_method',
    'selected_model', 'test_rows', 'test_covered_rows', 'test_coverage',
    'covered_directional_accuracy', 'covered_directional_majority_baseline',
    'covered_directional_accuracy_minus_baseline', 'covered_directional_balanced_accuracy',
    'covered_rmse', 'covered_baseline_rmse_train_mean', 'covered_r2', 'covered_correlation'
]
sel_cols = [c for c in sel_cols if c in large_selective.columns]
large_selective.sort_values(
    ['covered_directional_accuracy_minus_baseline', 'covered_directional_balanced_accuracy', 'covered_directional_accuracy'],
    ascending=[False, False, False]
)[sel_cols].head(20)
```
